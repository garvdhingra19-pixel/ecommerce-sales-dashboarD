```python
# Cell 2: Load and merge the raw datasets from the sidebar
import pandas as pd

print("Loading datasets...")
orders = pd.read_csv('olist_orders_dataset.csv')
order_items = pd.read_csv('olist_order_items_dataset.csv')
customers = pd.read_csv('olist_customers_dataset.csv')
products = pd.read_csv('olist_products_dataset.csv')

# Merge order items with main orders table
df = pd.merge(order_items, orders, on='order_id', how='inner')

# Merge with customer table
df = pd.merge(df, customers, on='customer_id', how='inner')

# Merge with product table
df = pd.merge(df, products, on='product_id', how='inner')

print(f"Successfully merged! Total rows in master dataset: {df.shape[0]:,}")
df.head(3)
```

```python
# Concept 1: Inspect missing values and filter for valid delivered orders
print("--- 1. Missing Values Before Cleaning ---")
print(df[['order_status', 'price', 'customer_id', 'product_category_name']].isnull().sum())

# Filter to keep only successfully delivered orders
df_cleaned = df[df['order_status'] == 'delivered'].copy()

# Drop rows where critical financial columns (price or freight) have missing values
df_cleaned = df_cleaned.dropna(subset=['price', 'freight_value'])

print(f"\nRows before cleaning: {df.shape[0]:,}")
print(f"Rows after cleaning: {df_cleaned.shape[0]:,}")
```

```python
# Define the exact important columns needed for your E-commerce Dashboard
essential_columns = [
    'order_id', 'customer_id', 'customer_unique_id',
    'order_status', 'order_purchase_timestamp',
    'price', 'freight_value',
    'product_category_name', 'customer_state'
]

# Keep only the important columns in your cleaned dataframe
df_final = df_cleaned[essential_columns].copy()

print(f"Original columns count: {df.shape[1]}")
print(f"Reduced essential columns count: {df_final.shape[1]}")
df_final.head(2)
```

```python
# Step 3: Feature Engineering and KPI Calculations

# 1. Create a total revenue column for each line item (Price + Freight)
df_final['total_revenue'] = df_final['price'] + df_final['freight_value']

# 2. Compute Core Financial and Volume KPIs
total_revenue = df_final['total_revenue'].sum()
total_orders = df_final['order_id'].nunique()
total_items = df_final['order_id'].count()
average_order_value = total_revenue / total_orders

# 3. Compute Repeat Customer Rate (using unique customer IDs)
customer_order_counts = df_final.groupby('customer_unique_id')['order_id'].nunique()
repeat_customers = (customer_order_counts > 1).sum()
total_unique_customers = customer_order_counts.count()
repeat_customer_rate = (repeat_customers / total_unique_customers) * 100

# 4. Print out formatted KPI results
print("=" * 40)
print("     E-COMMERCE DASHBOARD KEY METRICS     ")
print("=" * 40)
print(f"Total Revenue: R$ {total_revenue:,.2f}")
print(f"Total Unique Orders: {total_orders:,}")
print(f"Average Order Value (AOV): R$ {average_order_value:,.2f}")
print(f"Total Unique Customers: {total_unique_customers:,}")
print(f"Repeat Customer Rate: {repeat_customer_rate:.2f}%")
print("=" * 40)
```

```python
# Step 4: Monthly Sales Trend Analysis
import plotly.express as px

# 1. Extract Year-Month from the purchase timestamp
df_final['order_month'] = pd.to_datetime(df_final['order_purchase_timestamp']).dt.to_period('M').astype(str)

# 2. Group by month and sum the total revenue
monthly_sales = df_final.groupby('order_month')['total_revenue'].sum().reset_index()

# 3. Create an interactive line chart with Plotly
fig = px.line(
    monthly_sales,
    x='order_month',
    y='total_revenue',
    title='<b>Monthly E-Commerce Sales Trend</b>',
    labels={'order_month': 'Month', 'total_revenue': 'Total Revenue (R$)'},
    markers=True
)

fig.update_layout(xaxis_tickangle=-45, template='plotly_white')
fig.show()
```

```python
# Chart 2: Top 10 Product Categories by Total Revenue
import plotly.express as px

# 1. Group by product category and sum the revenue, then sort highest to lowest
category_sales = df_final.groupby('product_category_name')['total_revenue'].sum().reset_index()
top_categories = category_sales.sort_values(by='total_revenue', ascending=False).head(10)

# 2. Create an interactive horizontal bar chart with Plotly
fig2 = px.bar(
    top_categories,
    x='total_revenue',
    y='product_category_name',
    orientation='h',
    title='<b>Top 10 Product Categories by Revenue</b>',
    labels={'total_revenue': 'Total Revenue (R$)', 'product_category_name': 'Product Category'},
    text_auto=',.2f'
)

# 3. Clean up layout and display
fig2.update_layout(yaxis={'categoryorder':'total ascending'}, template='plotly_white')
fig2.show()
```

```python
# Sidelined Metric: Delivery Performance (On-Time vs. Late Deliveries)
import plotly.express as px
import pandas as pd

# 1. Load the original orders dataset to get estimated and actual delivery dates
orders_raw = pd.read_csv('olist_orders_dataset.csv')

# 2. Merge delivery date columns into our current cleaned dataframe
df_delivery = pd.merge(
    df_final,
    orders_raw[['order_id', 'order_estimated_delivery_date', 'order_delivered_customer_date']],
    on='order_id',
    how='left'
)

# 3. Convert date columns to true datetime objects
df_delivery['order_estimated_delivery_date'] = pd.to_datetime(df_delivery['order_estimated_delivery_date'])
df_delivery['order_delivered_customer_date'] = pd.to_datetime(df_delivery['order_delivered_customer_date'])

# 4. Calculate delivery delay in days (Actual minus Estimated)
df_delivery['delivery_delay_days'] = (
    df_delivery['order_delivered_customer_date'] - df_delivery['order_estimated_delivery_date']
).dt.days

# 5. Categorize each order's delivery outcome
def categorize_delivery(days):
    if pd.isna(days):
        return 'Data Missing'
    elif days > 0:
        return 'Late Delivery'
    else:
        return 'Early / On Time'

df_delivery['delivery_status'] = df_delivery['delivery_delay_days'].apply(categorize_delivery)

# 6. Count totals for each status and plot a Donut Chart
delivery_counts = df_delivery['delivery_status'].value_counts().reset_index()
delivery_counts.columns = ['delivery_status', 'order_count']

fig3 = px.pie(
    delivery_counts,
    names='delivery_status',
    values='order_count',
    hole=0.4,
    title='<b>Logistics Performance: On-Time vs. Late Deliveries</b>',
    color='delivery_status',
    color_discrete_map={'Early / On Time': '#2ca02c', 'Late Delivery': '#d62728', 'Data Missing': '#7f7f7f'}
)

fig3.update_traces(textposition='inside', textinfo='percent+label')
fig3.update_layout(template='plotly_white')
fig3.show()
```
